```python
import pandas as pd
data = pd.read_excel('all_data_M_2025.xlsx')
```

```python
print("Shape:", data.shape)
print("\nColumns:", data.columns.tolist())
print("\nFirst 5 rows:")
print(data.head())
```

```python
print(data['AREA_TYPE'].value_counts())
```

```python
for AREA_TYPE in [1,4, 6, 2, 3]:
    print(f"\nAREA_TYPE {AREA_TYPE}: ")
    print(data[data['AREA_TYPE'] == AREA_TYPE]['AREA_TITLE'].value_counts().head())
```

```python
data['A_MEAN'] = pd.to_numeric(data['A_MEAN'], errors='coerce')
data['A_MEDIAN'] = pd.to_numeric(data['A_MEDIAN'], errors='coerce')
```

```python
print(data[data['AREA_TYPE'] == 2]
      [['AREA_TITLE', 'OCC_TITLE', 'A_MEAN', 'A_MEDIAN']]
      .sort_values('A_MEAN', ascending=False).head(10))
```

```python
top_per_states = data[data['AREA_TYPE'] == 2].sort_values('A_MEAN', ascending=False).groupby('AREA_TITLE').first().reset_index()[['AREA_TITLE', 'OCC_TITLE', 'A_MEAN', "A_MEDIAN"]].sort_values('A_MEAN', ascending=False)
print(top_per_states)
```

```python
avg_salary = data[data['AREA_TYPE'] == 2].groupby('AREA_TITLE')['A_MEAN'].mean().sort_values(ascending=False)
print(avg_salary)
```

```python
avg_salary_data = avg_salary.reset_index()
avg_salary_data.columns = ['AREA_TITLE', 'A_MEAN']

state_abbrev = {'Alabama': 'AL', 'Alaska': 'AK', 'Arizona': 'AZ',
    'Arkansas': 'AR', 'California': 'CA', 'Colorado': 'CO',
    'Connecticut': 'CT', 'Delaware': 'DE', 'District of Columbia': 'DC',
    'Florida': 'FL', 'Georgia': 'GA', 'Hawaii': 'HI',
    'Idaho': 'ID', 'Illinois': 'IL', 'Indiana': 'IN',
    'Iowa': 'IA', 'Kansas': 'KS', 'Kentucky': 'KY',
    'Louisiana': 'LA', 'Maine': 'ME', 'Maryland': 'MD',
    'Massachusetts': 'MA', 'Michigan': 'MI', 'Minnesota': 'MN',
    'Mississippi': 'MS', 'Missouri': 'MO', 'Montana': 'MT',
    'Nebraska': 'NE', 'Nevada': 'NV', 'New Hampshire': 'NH',
    'New Jersey': 'NJ', 'New Mexico': 'NM', 'New York': 'NY',
    'North Carolina': 'NC', 'North Dakota': 'ND', 'Ohio': 'OH',
    'Oklahoma': 'OK', 'Oregon': 'OR', 'Pennsylvania': 'PA',
    'Rhode Island': 'RI', 'South Carolina': 'SC', 'South Dakota': 'SD',
    'Tennessee': 'TN', 'Texas': 'TX', 'Utah': 'UT',
    'Vermont': 'VT', 'Virginia': 'VA', 'Washington': 'WA',
    'West Virginia': 'WV', 'Wisconsin': 'WI', 'Wyoming': 'WY',
    'Puerto Rico': 'PR', 'Guam': 'GU', 'Virgin Islands': 'VI'
}

avg_salary_data['State Abbreviation'] = avg_salary_data['AREA_TITLE'].map(state_abbrev)
print(avg_salary_data)
```

```python
import plotly.express as px

figure = px.choropleth(
    avg_salary_data,
    locations = 'STATE_ABBREV',
    locationmode = 'USA-states',
    color = 'A_MEAN',
    scope = 'usa',
    title = 'Average Annual Wage by State (2025)',
    color_continuous_scale = 'Blues',
    labels={'A_MEAN': 'Average Annual Wage ($)'}
)

figure.show()
```
